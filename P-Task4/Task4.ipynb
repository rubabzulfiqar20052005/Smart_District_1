**Part1**

```python
products = [
    {"name": "Laptop", "price": 150000, "category": "Electronics"},
    {"name": "Mobile", "price": 80000, "category": "Electronics"},
    {"name": "Headphones", "price": 5000, "category": "Electronics"},
    {"name": "Tablet", "price": 60000, "category": "Electronics"},
    {"name": "Camera", "price": 120000, "category": "Electronics"},
    {"name": "Smart Watch", "price": 15000, "category": "Electronics"},
    {"name": "Shoes", "price": 8000, "category": "Fashion"},
    {"name": "Jacket", "price": 12000, "category": "Fashion"},
    {"name": "Watch", "price": 20000, "category": "Fashion"},
    {"name": "Bag", "price": 6000, "category": "Fashion"},
    {"name": "Sunglasses", "price": 4000, "category": "Fashion"},
    {"name": "Cap", "price": 1500, "category": "Fashion"},
    {"name": "Keyboard", "price": 3000, "category": "Electronics"},
    {"name": "Laptop", "price": 150000, "category": "Electronics"},
    {"name": "Mouse", "price": None, "category": "Electronics"},
    {"name": "Bottle", "price": 500, "category": None}
]
```

```python
cleaned_products = [
    p for p in products
    if p["price"] is not None and p["category"] is not None
]

cleaned_products = list({
    (p["name"], p["price"], p["category"]): p
    for p in cleaned_products
}.values())

print(cleaned_products)
```

```python
categories = {
    p["category"] for p in cleaned_products
}

grouped_products = {
    category: [p for p in cleaned_products if p["category"] == category]
    for category in categories
}

print(grouped_products)
```

```python
top_5_items = {
    category: sorted(items, key=lambda p: p["price"], reverse=True)[:5]
    for category, items in grouped_products.items()
}

for category, items in top_5_items.items():
    print("\nCategory:", category)

    for p in items:
        print(p["name"], "-", p["price"])
```

```python
summary = {
    category: {
        "total_products": len(items),
        "highest_price": max(p["price"] for p in items),
        "top_5": [p["name"] for p in items[:5]]
    }
    for category, items in top_5_items.items()
}

print(summary)
```

**Part2**

```python
employees = [
    {"name": "Ali", "score": 85, "experience": 3},
    {"name": "Sara", "score": 95, "experience": 4},
    {"name": "Ahmed", "score": 85, "experience": 5},
    {"name": "Ayesha", "score": 95, "experience": 4},
    {"name": "Usman", "score": 75, "experience": 2},
    {"name": "Hina", "score": 85, "experience": 5}
]
```

```python
ranked_employees = sorted(
    employees,
    key=lambda e: (-e["score"], -e["experience"], e["name"])
)

for e in ranked_employees:
    print(e)
```

```python
search_name = input("Enter employee name: ")

result = [
    e for e in ranked_employees
    if e["name"].lower() == search_name.lower()
]

if result:
    print("Employee found:", result[0])
else:
    print("Employee not found")
```

```python
employee_search = {
    e["name"].lower(): e
    for e in ranked_employees
}

name = input("Enter name to search: ").lower()

print(employee_search.get(name, "Employee not found"))
```

```python
hr_summary = {
    "total_employees": len(ranked_employees),
    "top_employee": ranked_employees[0]["name"],
    "rankings": [
        {
            "rank": i + 1,
            "name": e["name"],
            "score": e["score"],
            "experience": e["experience"]
        }
        for i, e in enumerate(ranked_employees)
    ]
}

print(hr_summary)
```
