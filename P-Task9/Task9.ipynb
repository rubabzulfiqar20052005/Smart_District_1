```python

class Book:
    def __init__(self, title):
        self.title = title
        self.available = True

    def __str__(self):
        return self.title
```

```python

class Member:
    def __init__(self, name, limit):
        self.name = name
        self.limit = limit
        self.borrowed_books = []

    def borrow_book(self, book):
        if book.available and len(self.borrowed_books) < self.limit:
            self.borrowed_books.append(book)
            book.available = False
            print(self.name, "borrowed", book.title)
        else:
            print("Book unavailable or borrowing limit reached!")

    def return_book(self, book, days_late):
        if book in self.borrowed_books:
            self.borrowed_books.remove(book)
            book.available = True
            fine = self.calculate_fine(days_late)
            print("Book returned:", book.title)
            print("Fine: Rs.", fine)
        else:
            print("This book was not borrowed by this member.")

    def calculate_fine(self, days_late):
        return max(0, days_late) * 10

    def __str__(self):
        return self.name + " | Borrowed books: " + str(len(self.borrowed_books))
```

```python

# Single inheritance
class RegularMember(Member):
    def __init__(self, name):
        super().__init__(name, 2)


# Single inheritance
class PremiumMember(Member):
    def __init__(self, name):
        super().__init__(name, 5)

    # Method overriding
    def calculate_fine(self, days_late):
        return max(0, days_late) * 5


# Multilevel inheritance: Member -> PremiumMember -> GoldMember
class GoldMember(PremiumMember):
    def __init__(self, name):
        super().__init__(name)
        self.limit = 10

    # Method overriding
    def calculate_fine(self, days_late):
        return max(0, days_late) * 2
```

```python

book1 = Book("Python Basics")
book2 = Book("Data Structures")
book3 = Book("Database Systems")

member1 = RegularMember("Ali")
member2 = PremiumMember("Sara")
member3 = GoldMember("Ahmed")
```

```python

member1.borrow_book(book1)
member1.borrow_book(book2)

member1.borrow_book(book3)

# Sara borrows a book
member2.borrow_book(book3)

# Book returned 4 days late
member2.return_book(book3, 4)
```

```python

book4 = Book("Artificial Intelligence")

member3.borrow_book(book4)
member3.return_book(book4, 4)
```

```python

print("----- Library Management Summary -----")

print(member1)
print(member2)
print(member3)

print("\nBook Status:")
print(book1, "- Available:", book1.available)
print(book2, "- Available:", book2.available)
print(book3, "- Available:", book3.available)
print(book4, "- Available:", book4.available)
```
