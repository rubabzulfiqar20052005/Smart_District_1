```python

logs = [
    "INFO: Server started",
    "ERROR: Database connection failed",
    "CRITICAL: Database is down",
    "WARNING: High memory usage",
    "CRITICAL: Payment service failed",
    "CRITICAL: Database is down",
    "INFO: Request completed",
    "CRITICAL: Payment service failed"
]

with open("server.log", "w") as file:
    file.write("\n".join(logs))

print("Log file created successfully!")
```

```python

def read_critical_errors(file_name):

    with open(file_name, "r") as file:
        for line in file:
            if "CRITICAL" in line:
                yield line.strip()
```

```python

from collections import defaultdict

error_groups = defaultdict(list)
error_count = 0

for error in read_critical_errors("server.log"):

    error_type = error.split(":")[1].strip()

    error_groups[error_type].append(error)

    error_count += 1

    print("Live Critical Error Count:", error_count)
```

```python

print("\nFINAL ERROR SUMMARY")
print("-------------------")

print("Total Critical Errors:", error_count)

for error_type, errors in error_groups.items():
    print("\nError Type:", error_type)
    print("Count:", len(errors))

    for error in errors:
        print("-", error)
```
