```python

orders = [
    {"customer": "Ali", "category": "Electronics", "amount": 5000},
    {"customer": "Sara", "category": "Clothing", "amount": 2000},
    {"customer": "Ahmed", "category": "Books", "amount": 500},
    {"customer": "Ayesha", "category": "Electronics", "amount": 3000},
    {"customer": "Usman", "category": "Clothing", "amount": 800}
]

print(orders)
```

```python

discounts = {
    "Electronics": 10,
    "Clothing": 20,
    "Books": 5
}

minimum_amount = 1000
```

```python

def apply_discount(order):
    discount = discounts.get(order["category"], 0)

    final_amount = order["amount"] * (1 - discount / 100)

    return {
        **order,
        "discount": discount,
        "final_amount": final_amount
    }

discounted_orders = list(map(apply_discount, orders))

print(discounted_orders)
```

```python

processed_orders = list(
    filter(lambda order: order["final_amount"] >= minimum_amount,
           discounted_orders)
)

print(processed_orders)
```

```python

total_revenue = sum(
    map(lambda order: order["final_amount"], processed_orders)
)

print("PROCESSED ORDER SUMMARY")
print("-----------------------")
print("Total raw orders:", len(orders))
print("Orders after filtering:", len(processed_orders))
print("Total revenue:", round(total_revenue, 2))

print("\nFinal Orders:")
print(processed_orders)
```
