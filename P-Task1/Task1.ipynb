```python
products = [
    {"name": "Laptop", "price": "50000", "category": "Electronics"},
    {"name": "Mouse", "price": 500, "category": "Electronics"},
    {"name": "Shirt", "price": "1500", "category": "Clothing"},
    {"name": "Laptop", "price": "50000", "category": "Electronics"},
    {"name": "Shoes", "price": 2500, "category": "Clothing"},
    {"name": "Keyboard", "price": None, "category": "Electronics"},
    {"name": "Bag", "price": "2000", "category": "Clothing"},
    {"name": "Phone", "price": "30000", "category": None}
]
```

```python
cleaned_products = [
    p for p in products
    if p["price"] is not None and p["category"] is not None
]

print(cleaned_products)
```

```python
cleaned_products = list({
    (p["name"], p["price"], p["category"]): p
    for p in cleaned_products
}.values())

print(cleaned_products)
```

```python
cleaned_products = [
    {**p, "price": float(p["price"])}
    for p in cleaned_products
]

print(cleaned_products)
```

```python
categories = {p["category"] for p in cleaned_products}

grouped_products = {
    category: [p for p in cleaned_products if p["category"] == category]
    for category in categories
}

print(grouped_products)
```

```python
for category, items in grouped_products.items():
    print("\nCategory:", category)
    print("Total products:", len(items))

    for p in items:
        print(p["name"], "- Rs.", p["price"])
```
