```python

import asyncio

async def get_sales():
    await asyncio.sleep(1)
    return {"sales": 5000, "orders": 50}

async def get_users():
    await asyncio.sleep(1)
    return {"users": 100, "new_users": 10}

async def get_products():
    await asyncio.sleep(1)
    return {"products": 25, "stock": 200}
```

```python

async def sync_data():
    sales_data, users_data, products_data = await asyncio.gather(
        get_sales(),
        get_users(),
        get_products()
    )

    combined_data = {
        **sales_data,
        **users_data,
        **products_data
    }

    return combined_data

data = await sync_data()

print("All data sources synced successfully!")
print(data)
```

```python

print("\n===== DATA SYNC DASHBOARD =====")

print("Total Sales: Rs.", data["sales"])
print("Total Orders:", data["orders"])
print("Total Users:", data["users"])
print("New Users:", data["new_users"])
print("Total Products:", data["products"])
print("Available Stock:", data["stock"])

print("===============================")
```
