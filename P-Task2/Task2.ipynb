```python
import pandas as pd
import os
import glob
```

```python
folder = "student_files"

os.makedirs(folder, exist_ok=True)
```

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [85, 35, 120]
}

df = pd.DataFrame(data)
df.to_csv("student_files/students1.csv", index=False)

data2 = {
    "Name": ["Ayesha", "Usman"],
    "Marks": [90, "abc"]
}

df2 = pd.DataFrame(data2)
df2.to_csv("student_files/students2.csv", index=False)
```

```python
all_students = []

files = glob.glob("student_files/*.csv")

for file in files:
    try:
        df = pd.read_csv(file)

        if "Name" not in df.columns or "Marks" not in df.columns:
            print("Invalid file:", file)
            continue

        all_students.extend(df.to_dict("records"))

    except Exception as e:
        print("File could not be read:", file)
```

```python
clean_students = []

for student in all_students:
    try:
        name = student["Name"]
        marks = float(student["Marks"])

        if pd.isna(name) or not str(name).strip():
            print("Invalid student name")
            continue

        if marks < 0 or marks > 100:
            print("Invalid marks for:", name)
            continue

        result = "Pass" if marks >= 50 else "Fail"

        clean_students.append({
            "Name": str(name).strip(),
            "Marks": marks,
            "Result": result
        })

    except (ValueError, TypeError):
        print("Invalid student data:", student)
```

```python
try:
    results = pd.DataFrame(clean_students)

    results.to_csv("clean_results.csv", index=False)

    print("Results saved successfully!")
    print(results)

except Exception as e:
    print("Results could not be saved:", e)
```
